```python
#!/usr/bin/env python3
from __future__ import annotations
 
import gc
import itertools
import logging
import os
import re
import sys
import time
import warnings
from concurrent.futures import ThreadPoolExecutor
from dataclasses import dataclass
from pathlib import Path
from typing import Dict, List, Optional, Sequence, Tuple
 
import numpy as np
import pandas as pd
 
warnings.filterwarnings("ignore")
 
# Use a non-interactive backend when not running inside a notebook.
import matplotlib
 
try:
    get_ipython  # type: ignore  # noqa: B018
except NameError:
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
 
from scipy import ndimage as ndi
from sklearn.base import BaseEstimator, RegressorMixin
from sklearn.compose import TransformedTargetRegressor
from sklearn.cross_decomposition import PLSRegression
from sklearn.decomposition import PCA
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import RidgeCV
from sklearn.model_selection import KFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
 
logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(message)s", datefmt="%H:%M:%S")
LOG = logging.getLogger("soil")
 
# ----------------------------------------------------------------------------
# Configuration
# ----------------------------------------------------------------------------
SUPPORT_MM = np.array([0.002, 0.0063, 0.02, 0.063, 0.2, 0.63, 2.0, 6.3, 20.0, 63.0, 200.0])
DEFAULT_SUB_COLS = ["0.002", "0.0063", "0.02", "0.063", "0.2", "0.63", "2.0", "6.3", "20", "63", "200"]
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"}
LOG_WIDTHS = np.diff(np.log10(SUPPORT_MM))  # ten interval widths in log10 units
 
 
@dataclass
class Config:
    data_dir: str = os.environ.get("SOIL_DATA_DIR", "/kaggle/input/competitions/soil-grain-size-from-photos")
    out_dir: str = os.environ.get("SOIL_OUT_DIR", "/kaggle/working" if os.path.isdir("/kaggle/working") else "./output")
    seed: int = 42
    n_folds: int = 5
    max_images_per_sample: int = 6      # cap to bound runtime for samples with many photos
    hand_img_size: int = 512            # size of the resized global view
    crop_size: int = 512                # native-scale crop size
    crop_short_side: int = 1024         # very large photos are reduced to this short side first
    n_workers: int = min(4, os.cpu_count() or 2)
    n_repeats: int = 3                  # CV repeats (small datasets only)
    knn_k: int = 4
    use_deep: bool = True               # pretrained embeddings (local DINOv2 or timm with internet)
    deep_model: str = "convnext_tiny"   # timm fallback
    deep_res: int = 448                 # multiple of 14 and 32; reduced to 224 on CPU
    deep_batch: int = 32
    use_lgbm: bool = True               # only used when there are >= 150 training samples
    time_budget_s: float = 40 * 60      # soft budget; optional stages are skipped beyond it
    corr_threshold: float = 0.95
 
 
CFG = Config()
T0 = time.time()
 
 
def time_left() -> float:
    return CFG.time_budget_s - (time.time() - T0)
 
 
# ----------------------------------------------------------------------------
# Metric and post-processing
# ----------------------------------------------------------------------------
def emd_log(y_true: np.ndarray, y_pred: np.ndarray) -> float:
    """Mean log-weighted EMD. Both arrays have shape (n, 11), values in percent."""
    y_true = np.asarray(y_true, dtype=np.float64)
    y_pred = np.asarray(y_pred, dtype=np.float64)
    return float(np.mean(np.abs(y_true[:, :-1] - y_pred[:, :-1]) @ LOG_WIDTHS))
 
 
def postprocess(pred10: np.ndarray) -> np.ndarray:
    """Turn raw predictions of the first ten points into valid (n, 11) curves."""
    p = np.clip(np.asarray(pred10, dtype=np.float64), 0.0, 100.0)
    p = np.maximum.accumulate(p, axis=1)
    return np.hstack([p, np.full((p.shape[0], 1), 100.0)])
 
 
# ----------------------------------------------------------------------------
# Data discovery and loading
# ----------------------------------------------------------------------------
def resolve_data_dir(path: str) -> Path:
    """Return an existing dataset directory, searching /kaggle/input if needed."""
    p = Path(path)
    if p.is_dir():
        return p
    root = Path("/kaggle/input")
    if root.is_dir():
        hits = [d for d in root.rglob("*") if d.is_dir() and "soil" in d.name.lower() and "grain" in d.name.lower()]
        if hits:
            LOG.info("Configured data dir missing, using %s", hits[0])
            return hits[0]
        children = [d for d in root.iterdir() if d.is_dir()]
        if len(children) == 1:
            return children[0]
    raise FileNotFoundError(f"Dataset directory not found: {path}")
 
 
def scan_files(root: Path) -> Tuple[List[Path], List[Path]]:
    csvs, imgs = [], []
    for dirpath, _, files in os.walk(root):
        for f in files:
            ext = os.path.splitext(f)[1].lower()
            fp = Path(dirpath) / f
            if ext == ".csv":
                csvs.append(fp)
            elif ext in IMG_EXT:
                imgs.append(fp)
    return sorted(csvs), sorted(imgs)
 
 
def support_column_map(columns: Sequence[str]) -> Dict[str, int]:
    """Map column names that parse as a support diameter to their support index."""
    out: Dict[str, int] = {}
    for c in columns:
        try:
            v = float(str(c).strip())
        except ValueError:
            continue
        idx = np.where(np.isclose(SUPPORT_MM, v, rtol=1e-6))[0]
        if idx.size:
            out[c] = int(idx[0])
    return out
 
 
def read_csv_safe(path: Path) -> Optional[pd.DataFrame]:
    for enc in ("utf-8", "utf-8-sig", "latin-1"):
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception:  # noqa: BLE001
            continue
    LOG.warning("Could not read %s", path)
    return None
 
 
def classify_csvs(csvs: List[Path]) -> Dict[str, Optional[Tuple[Path, pd.DataFrame]]]:
    """Identify train, test and sample submission files by name and content."""
    found: Dict[str, Optional[Tuple[Path, pd.DataFrame]]] = {"train": None, "test": None, "sub": None}
    others: List[Tuple[Path, pd.DataFrame]] = []
    for p in csvs:
        df = read_csv_safe(p)
        if df is None or df.empty:
            continue
        name = p.stem.lower()
        n_sup = len(support_column_map(df.columns))
        if n_sup >= 8:
            const_rows = df[[c for c in df.columns if c in support_column_map(df.columns)]].nunique().max() <= 1
            if "sub" in name or "sample" in name or const_rows:
                found["sub"] = (p, df)
            elif found["train"] is None or "train" in name:
                found["train"] = (p, df)
        elif "test" in name and found["test"] is None:
            found["test"] = (p, df)
        else:
            others.append((p, df))
    if found["test"] is None:
        for p, df in others:  # metadata file without labels that is not obviously train
            if "train" not in p.stem.lower():
                found["test"] = (p, df)
                break
    return found
 
 
def id_column(df: pd.DataFrame) -> str:
    for c in df.columns:
        if str(c).lower() in ("sample_id", "sampleid", "id"):
            return c
    return df.columns[0]
 
 
def prepare_targets(train: pd.DataFrame, sid: str) -> Tuple[pd.DataFrame, np.ndarray]:
    """Validate labels: numeric coercion, unit check, interpolation, monotone repair."""
    cmap = support_column_map(train.columns)
    cols = sorted(cmap, key=lambda c: cmap[c])
    if len(cols) < 11:
        raise ValueError(f"Train file has {len(cols)} of 11 support columns")
    Y = train[cols].apply(pd.to_numeric, errors="coerce").astype(np.float64)
    if np.nanmax(Y.values) <= 1.0 + 1e-9:  # fractions instead of percent
        Y = Y * 100.0
    Y = Y.interpolate(axis=1, limit_direction="both")
    keep = Y.notna().all(axis=1) & train[sid].notna()
    dropped = int((~keep).sum())
    if dropped:
        LOG.info("Dropped %d train rows with unusable labels", dropped)
    train = train.loc[keep].reset_index(drop=True)
    arr = np.clip(Y.loc[keep].values, 0.0, 100.0)
    arr = np.maximum.accumulate(arr, axis=1)
    arr[:, -1] = 100.0
    return train, arr
 
 
def _tokens(text: str) -> Tuple[str, ...]:
    """Case-insensitive token tuple; letters and digits split, leading zeros removed."""
    return tuple(str(int(t)) if t.isdigit() else t for t in re.findall(r"[a-z]+|\d+", str(text).lower()))
 
 
def _rel_parts(p: Path, root: Optional[Path]) -> List[str]:
    """Path components (stem first, then parent folders) relative to the dataset root."""
    try:
        parts = list(p.relative_to(root).parts) if root is not None else list(p.parts)
    except ValueError:
        parts = list(p.parts)
    return [Path(parts[-1]).stem] + list(reversed(parts[:-1]))
 
 
def _is_text_col(s: pd.Series) -> bool:
    return pd.api.types.is_object_dtype(s) or pd.api.types.is_string_dtype(s)
 
 
def map_images_to_samples(
    ids: Sequence[str],
    img_paths: List[Path],
    meta: Optional[pd.DataFrame],
    sid: Optional[str],
    root: Optional[Path] = None,
    split_hint: Optional[str] = None,
) -> Dict[str, List[Path]]:
    """Assign image files to sample ids.
 
    Strategy 1: CSV columns that reference image files (basename, with or without
    extension, one or several names per cell).
    Strategy 2: token matching of the sample id against file name, then folder names,
    tolerant to case, separators and zero padding (TRAIN_01 == train-1 == Train1).
    """
    ids = [str(i) for i in ids]
    result: Dict[str, List[Path]] = {i: [] for i in ids}
    assigned: set = set()
 
    # Restrict to the train or test image pool when folder names make that possible.
    pool = list(img_paths)
    if split_hint:
        low = [(p, [x.lower() for x in _rel_parts(p, root)]) for p in pool]
        has_tr = any(any("train" in x for x in parts) for _, parts in low)
        has_te = any(any("test" in x for x in parts) for _, parts in low)
        if has_tr and has_te:
            sub = [p for p, parts in low if any(split_hint in x for x in parts)]
            if sub:
                pool = sub
 
    by_name: Dict[str, List[Path]] = {}
    by_stem: Dict[str, List[Path]] = {}
    for p in pool:
        by_name.setdefault(p.name.lower(), []).append(p)
        by_stem.setdefault(p.stem.lower(), []).append(p)
 
    def lookup(piece: str) -> List[Path]:
        piece = piece.strip().strip("'\"[]").replace("\\", "/")
        base = piece.split("/")[-1].lower()
        cand = by_name.get(base) or by_stem.get(base) or by_stem.get(os.path.splitext(base)[0]) or []
        if len(cand) > 1 and "/" in piece:
            tail = piece.lower().lstrip("./")
            narrowed = [c for c in cand if c.as_posix().lower().endswith(tail)]
            cand = narrowed or cand
        return cand
 
    # Strategy 1: image references inside the CSV.
    if meta is not None and sid is not None and sid in meta.columns:
        for col in meta.columns:
            if col == sid or not _is_text_col(meta[col]):
                continue
            cells = meta[col].fillna("").astype(str)
            pieces = cells.map(lambda v: [x for x in re.split(r"[;,|\n]+", v) if x.strip()])
            hit = np.mean([any(lookup(x) for x in pc) for pc in pieces]) if len(pieces) else 0
            if hit < 0.5:
                continue
            for i, pc in zip(meta[sid].astype(str), pieces):
                if i not in result:
                    continue
                for x in pc:
                    for p in lookup(x):
                        if p not in result[i]:
                            result[i].append(p)
                            assigned.add(p)
 
    # Strategy 2: match ids to path tokens for images not yet assigned.
    key_to_id: Dict[Tuple[str, ...], Optional[str]] = {}
    for i in ids:
        k = _tokens(i)
        if not k:
            continue
        key_to_id[k] = i if k not in key_to_id else None  # None marks ambiguous keys
    max_len = max((len(k) for k in key_to_id), default=1)
    for p in pool:
        if p in assigned:
            continue
        found = None
        for comp in _rel_parts(p, root):  # file name first, then folders
            t = _tokens(comp)
            best = None
            for length in range(min(len(t), max_len), 0, -1):
                for a in range(0, len(t) - length + 1):
                    hit_id = key_to_id.get(t[a:a + length])
                    if hit_id is not None:
                        best = hit_id
                        break
                if best is not None:
                    break
            if best is not None:
                found = best
                break
        if found is not None and p not in result[found]:
            result[found].append(p)
    return result
 
 
def log_mapping_diagnostics(name: str, ids: Sequence[str], img_paths: List[Path], root: Path,
                            meta: Optional[pd.DataFrame]) -> None:
    """Log compact evidence to debug an image/id mismatch."""
    LOG.info("%s ids (first 5): %s", name, list(ids)[:5])
    LOG.info("Image paths (first 6): %s", [p.relative_to(root).as_posix() for p in img_paths[:6]])
    if meta is not None:
        LOG.info("%s CSV columns: %s", name, list(meta.columns)[:15])
        LOG.info("%s CSV first row: %s", name, meta.iloc[0].astype(str).tolist()[:15])
 
 
# ----------------------------------------------------------------------------
# Image loading and handcrafted features
# ----------------------------------------------------------------------------
def load_rgb(path: Path, size: int) -> Optional[Tuple[np.ndarray, Tuple[int, int]]]:
    """Load an image as uint8 RGB resized to (size, size). Returns None on failure."""
    from PIL import Image
 
    try:
        with Image.open(path) as im:
            orig = im.size
            im.draft("RGB", (size * 2, size * 2))  # fast JPEG downscaled decode
            im = im.convert("RGB").resize((size, size), Image.BILINEAR)
            return np.asarray(im, dtype=np.uint8), orig
    except Exception:  # noqa: BLE001
        return None
 
 
def load_views(path: Path) -> Optional[Tuple[np.ndarray, List[np.ndarray]]]:
    """Return a global 512x512 view and native-scale crops of one photograph.
 
    Grain size in pixels is only comparable across photos when the pixel scale is
    preserved, so besides the resized global view three crops are cut from the image
    at (nearly) native resolution. Very large photos are first reduced so that the
    short side is at most crop_short_side.
    """
    from PIL import Image
 
    size, crop, short = CFG.hand_img_size, CFG.crop_size, CFG.crop_short_side
    try:
        with Image.open(path) as im:
            im.draft("RGB", (short * 2, short * 2))
            im = im.convert("RGB")
            w, h = im.size
            glob = np.asarray(im.resize((size, size), Image.BILINEAR), dtype=np.uint8)
            s = min(w, h)
            if s > short:
                f = short / s
                im = im.resize((max(crop, round(w * f)), max(crop, round(h * f))), Image.BILINEAR)
            w, h = im.size
            if min(w, h) < crop:
                im = im.resize((max(crop, w), max(crop, h)), Image.BILINEAR)
                w, h = im.size
            crops = []
            for fx, fy in ((0.5, 0.5), (0.25, 0.25), (0.75, 0.75)):
                x0 = int(np.clip(fx * w - crop / 2, 0, w - crop))
                y0 = int(np.clip(fy * h - crop / 2, 0, h - crop))
                crops.append(np.asarray(im.crop((x0, y0, x0 + crop, y0 + crop)), dtype=np.uint8))
            return glob, crops
    except Exception:  # noqa: BLE001
        return None
 
 
def _block_mean(a: np.ndarray, f: int) -> np.ndarray:
    h, w = (a.shape[0] // f) * f, (a.shape[1] // f) * f
    return a[:h, :w].reshape(h // f, f, w // f, f).mean(axis=(1, 3))
 
 
def handcrafted_features(rgb: np.ndarray) -> Dict[str, float]:
    """Colour, gradient, blob-scale, local variance and frequency features of one view.
 
    Contrast-dependent quantities are complemented by normalised versions (divided by
    the image contrast) so that the model sees texture scale rather than lighting.
    """
    f: Dict[str, float] = {}
    x = rgb.astype(np.float32) / 255.0
    gray = x @ np.array([0.299, 0.587, 0.114], dtype=np.float32)
    gstd = float(gray.std()) + 1e-6
 
    for k, name in enumerate("rgb"):
        ch = x[..., k]
        f[f"{name}_mean"], f[f"{name}_std"] = float(ch.mean()), float(ch.std())
    mx, mn = x.max(axis=2), x.min(axis=2)
    sat = np.where(mx > 1e-6, (mx - mn) / np.maximum(mx, 1e-6), 0)
    f["sat_mean"], f["sat_std"], f["val_mean"] = float(sat.mean()), float(sat.std()), float(mx.mean())
    f["rg_ratio"] = float(x[..., 0].mean() / (x[..., 1].mean() + 1e-6))
    f["bg_ratio"] = float(x[..., 2].mean() / (x[..., 1].mean() + 1e-6))
    for q, v in zip((1, 5, 25, 50, 75, 95, 99), np.percentile(gray, (1, 5, 25, 50, 75, 95, 99))):
        f[f"gray_p{q}"] = float(v)
    g = gray - gray.mean()
    f["gray_skew"], f["gray_kurt"] = float((g ** 3).mean() / gstd ** 3), float((g ** 4).mean() / gstd ** 4)
    hist, _ = np.histogram(gray, bins=12, range=(0, 1), density=True)
    for k, v in enumerate(hist):
        f[f"gray_h{k}"] = float(v)
 
    # Gradient magnitude at several scales (fine texture -> coarse structure).
    for scale in (1, 2, 4, 8):
        a = gray if scale == 1 else _block_mean(gray, scale)
        gx, gy = ndi.sobel(a, axis=1) / 8.0, ndi.sobel(a, axis=0) / 8.0
        mag = np.hypot(gx, gy)
        f[f"grad{scale}_mean"], f[f"grad{scale}_std"] = float(mag.mean()), float(mag.std())
        f[f"gradn{scale}"] = float(mag.mean() / gstd)
        f[f"edge{scale}"] = float((mag > 0.05).mean())
        f[f"lapn{scale}"] = float(np.sqrt(ndi.laplace(a).var()) / gstd)
 
    # Difference-of-Gaussians energy per blob scale (grain-size proxy), raw and profile.
    prev, dog = gray, []
    for sigma in (1, 2, 4, 8, 16, 32):
        cur = ndi.gaussian_filter(gray, sigma)
        dog.append(float((prev - cur).std()))
        prev = cur
    tot = sum(dog) + 1e-9
    for sigma, v in zip((1, 2, 4, 8, 16, 32), dog):
        f[f"dog{sigma}_std"], f[f"dogn{sigma}"] = v, v / tot
    f["dog_centroid"] = float(sum(np.log2(s) * v for s, v in zip((1, 2, 4, 8, 16, 32), dog)) / tot)
 
    # Local standard deviation at several window sizes.
    for w in (3, 7, 15, 31):
        m1 = ndi.uniform_filter(gray, w)
        m2 = ndi.uniform_filter(gray * gray, w)
        ls = np.sqrt(np.maximum(m2 - m1 * m1, 0))
        f[f"lstd{w}_mean"], f[f"lstdn{w}"] = float(ls.mean()), float(ls.mean() / gstd)
 
    # Radial power spectrum (log-spaced bands), normalised.
    win = np.outer(np.hanning(gray.shape[0]), np.hanning(gray.shape[1])).astype(np.float32)
    ps = np.abs(np.fft.fftshift(np.fft.fft2((gray - gray.mean()) * win))) ** 2
    cy, cx = np.array(ps.shape) // 2
    yy, xx = np.indices(ps.shape)
    rr = np.hypot(yy - cy, xx - cx)
    edges = np.geomspace(2, rr.max() / 1.5, 13)
    psd_tot = ps.sum() + 1e-9
    band = []
    for k in range(12):
        m = (rr >= edges[k]) & (rr < edges[k + 1])
        band.append(float(ps[m].sum() / psd_tot) if m.any() else 0.0)
        f[f"psd{k}"] = float(np.log10(band[-1] + 1e-12))
    band_a = np.array(band) + 1e-12
    f["psd_centroid"] = float((np.arange(12) * band_a).sum() / band_a.sum())
    # Spectral slope: coarse (large-grained) soils have shallower spectra.
    f["psd_slope"] = float(np.polyfit(np.log10(edges[:12]), np.log10(band_a), 1)[0])
    return f
 
 
def image_features(path: Path) -> Optional[Dict[str, float]]:
    """Features of the global view (g_) and of the mean over native-scale crops (c_)."""
    views = load_views(path)
    if views is None:
        return None
    try:
        glob, crops = views
        out = {f"g_{k}": v for k, v in handcrafted_features(glob).items()}
        cf = [handcrafted_features(c) for c in crops]
        for k in cf[0]:
            out[f"c_{k}"] = float(np.mean([d[k] for d in cf]))
        return out
    except Exception:  # noqa: BLE001
        return None
 
 
def _sample_jobs(sample_imgs: Dict[str, List[Path]]) -> List[Tuple[str, Path]]:
    jobs: List[Tuple[str, Path]] = []
    for sid, paths in sample_imgs.items():
        if paths:
            ordered = sorted(paths)
            step = max(1, len(ordered) // CFG.max_images_per_sample)
            jobs += [(sid, p) for p in ordered[::step][: CFG.max_images_per_sample]]
    return jobs
 
 
def aggregate_samples(rows: List[Tuple[str, Dict[str, float]]]) -> pd.DataFrame:
    """Mean of per-image features per sample (means are far less noisy than stds for tiny n)."""
    if not rows:
        return pd.DataFrame()
    df = pd.DataFrame([r for _, r in rows]).astype(np.float32)
    df["__id"] = [i for i, _ in rows]
    return df.groupby("__id").mean()
 
 
def build_handcrafted(sample_imgs: Dict[str, List[Path]]) -> pd.DataFrame:
    jobs = _sample_jobs(sample_imgs)
    LOG.info("Handcrafted features for %d images", len(jobs))
    with ThreadPoolExecutor(CFG.n_workers) as ex:
        res = list(ex.map(image_features, [p for _, p in jobs]))
    rows = [(sid, r) for (sid, _), r in zip(jobs, res) if r is not None]
    if len(rows) < len(jobs):
        LOG.info("%d images could not be read", len(jobs) - len(rows))
    return aggregate_samples(rows)
 
 
# ----------------------------------------------------------------------------
# pretrained embeddings
# ----------------------------------------------------------------------------
def find_local_deep_dir() -> Optional[Path]:
    """Locate an attached DINOv2 model (HuggingFace format) under /kaggle/input."""
    env = os.environ.get("SOIL_DEEP_DIR")
    if env and Path(env).is_dir():
        return Path(env)
    root = Path("/kaggle/input")
    if not root.is_dir():
        return None
    cands = [p.parent for p in root.rglob("config.json") if "dinov2" in p.as_posix().lower()]
    if not cands:
        return None
 
    def rank(p: Path) -> Tuple[int, int]:
        s = p.as_posix().lower()
        return (0 if "base" in s else 1 if "small" in s else 2, len(s))
 
    return sorted(cands, key=rank)[0]
 
 
def build_deep(sample_imgs: Dict[str, List[Path]]) -> Optional[pd.DataFrame]:
    """Mean-pooled embeddings per sample. Returns None when no model can be loaded."""
    if not CFG.use_deep or time_left() < 10 * 60:
        return None
    try:
        import torch
        from torch.utils.data import DataLoader, Dataset
 
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        res = CFG.deep_res if device.type == "cuda" else 224
        local = find_local_deep_dir()
        if local is not None:
            from transformers import AutoModel
 
            model, kind = AutoModel.from_pretrained(str(local)), "hf"
            mean_std = ((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))
            LOG.info("Embedding model: local %s on %s", local.name, device.type)
        else:
            import timm
 
            model, kind = timm.create_model(CFG.deep_model, pretrained=True, num_classes=0), "timm"
            pc = getattr(model, "pretrained_cfg", {}) or {}
            mean_std = (pc.get("mean", (0.485, 0.456, 0.406)), pc.get("std", (0.229, 0.224, 0.225)))
            LOG.info("Embedding model: timm %s on %s", CFG.deep_model, device.type)
        model = model.to(device).eval()
        mean = torch.tensor(mean_std[0], device=device).view(1, 3, 1, 1)
        std = torch.tensor(mean_std[1], device=device).view(1, 3, 1, 1)
        jobs = _sample_jobs(sample_imgs)
 
        class DS(Dataset):
            def __len__(self) -> int:
                return len(jobs)
 
            def __getitem__(self, i: int):
                loaded = load_rgb(jobs[i][1], res)
                if loaded is None:
                    return torch.zeros(3, res, res, dtype=torch.uint8), 0
                return torch.from_numpy(loaded[0]).permute(2, 0, 1).contiguous(), 1
 
        loader = DataLoader(DS(), batch_size=CFG.deep_batch, num_workers=2, shuffle=False)
        feats, valid = [], []
        with torch.inference_mode():
            for xb, vb in loader:
                xb = (xb.to(device).float() / 255.0 - mean) / std
                with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=device.type == "cuda"):
                    if kind == "hf":
                        h = model(pixel_values=xb).last_hidden_state
                        out = torch.cat([h[:, 0], h[:, 1:].mean(dim=1)], dim=1)
                    else:
                        out = model(xb)
                feats.append(out.float().cpu().numpy())
                valid.append(vb.numpy())
        fa, va = np.vstack(feats), np.concatenate(valid).astype(bool)
        rows = [(jobs[i][0], dict(enumerate(fa[i]))) for i in range(len(jobs)) if va[i]]
        out_df = aggregate_samples(rows)
        out_df.columns = [f"dp_{c}" for c in out_df.columns]
        del model
        gc.collect()
        if device.type == "cuda":
            torch.cuda.empty_cache()
        LOG.info("Embeddings: %d dims", out_df.shape[1])
        return out_df
    except Exception as exc:  # noqa: BLE001
        LOG.info("Embeddings unavailable (%s), using handcrafted features only", type(exc).__name__)
        return None
 
 
# ----------------------------------------------------------------------------
# EDA
# ----------------------------------------------------------------------------
def run_eda(Y: np.ndarray, ids: pd.Series, sample_imgs: Dict[str, List[Path]], feats: pd.DataFrame, out: Path) -> None:
    """Save EDA figures and log concise statistics."""
    out.mkdir(parents=True, exist_ok=True)
    labels = DEFAULT_SUB_COLS
    n = len(Y)
    LOG.info("EDA: %d training samples, %d support points", n, Y.shape[1])
    desc = pd.DataFrame(Y, columns=labels).describe().T[["mean", "std", "min", "50%", "max"]].round(2)
    LOG.info("Cumulative %% statistics per diameter:\n%s", desc.to_string())
 
    # Derived soil descriptors by log-linear interpolation of the cumulative curve.
    logx = np.log10(SUPPORT_MM)
 
    def dq(q: float) -> np.ndarray:
        return np.array([10 ** np.interp(q, row, logx) for row in Y])
 
    d10, d50, d60 = dq(10), dq(50), dq(60)
    frac = {
        "clay <0.002": Y[:, 0], "silt 0.002-0.063": Y[:, 3] - Y[:, 0],
        "sand 0.063-2": Y[:, 6] - Y[:, 3], "gravel 2-63": Y[:, 9] - Y[:, 6],
    }
    LOG.info("Median fractions (%%): %s", {k: round(float(np.median(v)), 1) for k, v in frac.items()})
    LOG.info("Median D50 = %.3f mm, share of samples with D50 > 2 mm = %.1f %%", np.median(d50), 100 * np.mean(d50 > 2))
 
    fig, ax = plt.subplots(2, 3, figsize=(18, 10))
    ax[0, 0].boxplot([Y[:, i] for i in range(11)], labels=labels)
    ax[0, 0].set_title("Cumulative % per support diameter")
    ax[0, 0].tick_params(axis="x", rotation=60)
    show = np.random.default_rng(0).choice(n, size=min(60, n), replace=False)
    for i in show:
        ax[0, 1].plot(SUPPORT_MM, Y[i], alpha=0.3, color="tab:blue")
    ax[0, 1].plot(SUPPORT_MM, np.median(Y, axis=0), color="red", lw=2, label="median")
    ax[0, 1].set_xscale("log")
    ax[0, 1].set_title("Grading curves")
    ax[0, 1].set_xlabel("diameter [mm]")
    ax[0, 1].legend()
    im = ax[0, 2].imshow(np.corrcoef(Y[:, :10].T), vmin=-1, vmax=1, cmap="coolwarm")
    ax[0, 2].set_title("Target correlation")
    fig.colorbar(im, ax=ax[0, 2])
    ax[1, 0].hist(np.log10(d50), bins=30, color="tab:green")
    ax[1, 0].set_title("log10 D50 [mm]")
    ax[1, 1].scatter(np.log10(d10), np.log10(d60), s=8, alpha=0.5)
    ax[1, 1].set_title("log10 D10 vs log10 D60")
    counts = np.array([len(sample_imgs.get(str(i), [])) for i in ids])
    ax[1, 2].hist(counts, bins=np.arange(counts.max() + 2) - 0.5)
    ax[1, 2].set_title("Images per training sample")
    plt.tight_layout()
    fig.savefig(out / "eda_targets.png", dpi=110)
    plt.show()
    plt.close(fig)
 
    if not feats.empty:
        aligned = feats.reindex(ids.astype(str).values)
        d50l = np.log10(d50)
        corr = aligned.apply(lambda c: c.corr(pd.Series(d50l, index=c.index)) if c.std() > 0 else np.nan).dropna()
        top = corr.reindex(corr.abs().sort_values(ascending=False).index).head(15)
        LOG.info("Top feature correlations with log10 D50:\n%s", top.round(3).to_string())
        fig, ax = plt.subplots(1, 2, figsize=(16, 5))
        top[::-1].plot.barh(ax=ax[0])
        ax[0].set_title("Feature vs log10 D50 (Pearson)")
        X = aligned.fillna(aligned.median()).values
        X = StandardScaler().fit_transform(X[:, aligned.std().values > 0])
        p2 = PCA(2, random_state=0).fit_transform(X)
        sc = ax[1].scatter(p2[:, 0], p2[:, 1], c=d50l, cmap="viridis", s=12)
        fig.colorbar(sc, ax=ax[1], label="log10 D50")
        ax[1].set_title("PCA of features")
        plt.tight_layout()
        fig.savefig(out / "eda_features.png", dpi=110)
        plt.show()
        plt.close(fig)
 
    # Example photos ordered from fine to coarse D50.
    order = np.argsort(d50)
    picks = [i for i in order[np.linspace(0, n - 1, 6).astype(int)]]
    fig, ax = plt.subplots(2, 6, figsize=(24, 7))
    for col, i in enumerate(picks):
        paths = sample_imgs.get(str(ids.iloc[i]), [])
        loaded = load_rgb(paths[0], 384) if paths else None
        if loaded is not None:
            ax[0, col].imshow(loaded[0])
        ax[0, col].axis("off")
        ax[0, col].set_title(f"D50={d50[i]:.3f} mm")
        ax[1, col].plot(SUPPORT_MM, Y[i], marker="o")
        ax[1, col].set_xscale("log")
        ax[1, col].set_ylim(0, 100)
    plt.tight_layout()
    fig.savefig(out / "eda_examples.png", dpi=90)
    plt.show()
    plt.close(fig)
 
 
# ----------------------------------------------------------------------------
# Feature filtering and modelling
# ----------------------------------------------------------------------------
def filter_features(train_X: pd.DataFrame, corr_threshold: float) -> List[str]:
    """Drop near-constant and highly correlated columns (unsupervised, leak-free)."""
    X = train_X.replace([np.inf, -np.inf], np.nan)
    X = X.loc[:, X.notna().mean() > 0.5]
    X = X.loc[:, X.std(skipna=True) > 1e-8]
    corr = X.fillna(X.median()).corr().abs().values
    keep = np.ones(X.shape[1], dtype=bool)
    for i in range(X.shape[1]):
        if keep[i]:
            dup = np.where(corr[i, i + 1:] > corr_threshold)[0] + i + 1
            keep[dup] = False
    cols = list(X.columns[keep])
    LOG.info("Feature filter: %d -> %d columns", train_X.shape[1], len(cols))
    return cols
 
 
class KNNMedian(BaseEstimator, RegressorMixin):
    """k-nearest-neighbour regressor returning the per-column median of the neighbours.
 
    The median is the L1-optimal aggregate, which matches the EMD metric, and the
    prediction is always a real grading curve, so it stays plausible with tiny data.
    """
 
    def __init__(self, k: int = 4):
        self.k = k
 
    def fit(self, X, y):
        self.X_ = np.asarray(X, dtype=np.float64)
        self.y_ = np.asarray(y, dtype=np.float64)
        return self
 
    def predict(self, X):
        X = np.asarray(X, dtype=np.float64)
        d = ((X[:, None, :] - self.X_[None, :, :]) ** 2).sum(axis=-1)
        idx = np.argsort(d, axis=1)[:, : min(self.k, len(self.X_))]
        return np.median(self.y_[idx], axis=1)
 
 
def _prep() -> List[Tuple[str, object]]:
    return [("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())]
 
 
def make_model(kind: str):
    """Low-variance regressors suited to a few dozen training samples."""
    if kind == "const":
        return DummyRegressor(strategy="median")
    if kind == "knn":
        return Pipeline(_prep() + [("pca", PCA(n_components=0.9, svd_solver="full", random_state=CFG.seed)),
                                   ("m", KNNMedian(CFG.knn_k))])
    if kind == "pls":
        return Pipeline(_prep() + [("m", PLSRegression(n_components=2, scale=False))])
    if kind == "ridge":
        return Pipeline(_prep() + [("m", RidgeCV(alphas=np.logspace(0, 5, 21), alpha_per_target=True))])
    if kind == "tpca":
        # Predict the first three principal components of the curves, then reconstruct.
        reg = Pipeline(_prep() + [("m", RidgeCV(alphas=np.logspace(0, 5, 21), alpha_per_target=True))])
        return TransformedTargetRegressor(regressor=reg, transformer=PCA(n_components=3), check_inverse=False)
    if kind == "et":
        return Pipeline([("imp", SimpleImputer(strategy="median")),
                         ("m", ExtraTreesRegressor(n_estimators=300, min_samples_leaf=2, max_features=0.3,
                                                   n_jobs=-1, random_state=CFG.seed))])
    if kind == "lgbm":
        from lightgbm import LGBMRegressor
        from sklearn.multioutput import MultiOutputRegressor
 
        return Pipeline([("imp", SimpleImputer(strategy="median")),
                         ("m", MultiOutputRegressor(LGBMRegressor(
                             objective="l1", n_estimators=300, learning_rate=0.04, num_leaves=15,
                             min_child_samples=5, subsample=0.8, subsample_freq=1, colsample_bytree=0.5,
                             n_jobs=-1, random_state=CFG.seed, verbose=-1)))])
    raise ValueError(kind)
 
 
ModelSpec = Tuple[str, str, str]  # (name, kind, feature group)
 
 
def cross_validate(specs: List[ModelSpec], groups: Dict[str, np.ndarray], Y: np.ndarray) -> Dict[str, np.ndarray]:
    """Repeated K-fold out-of-fold predictions, shape (repeats, n, 10) per model."""
    n = len(Y)
    repeats = CFG.n_repeats if n < 200 else 1
    splits = min(CFG.n_folds, n)
    oof = {name: np.zeros((repeats, n, 10)) for name, _, _ in specs}
    failed: List[str] = []
    for r in range(repeats):
        kf = KFold(splits, shuffle=True, random_state=CFG.seed + r)
        for tr, va in kf.split(np.arange(n)):
            for name, kind, grp in specs:
                if name in failed:
                    continue
                try:
                    X = groups[grp]
                    oof[name][r, va] = make_model(kind).fit(X[tr], Y[tr, :10]).predict(X[va])
                except Exception as exc:  # noqa: BLE001
                    LOG.info("Model %s failed (%s), dropped", name, type(exc).__name__)
                    failed.append(name)
    for name in failed:
        oof.pop(name, None)
    return oof
 
 
def oof_score(pred: np.ndarray, Y: np.ndarray) -> float:
    return float(np.mean([emd_log(Y, postprocess(pred[r])) for r in range(pred.shape[0])]))
 
 
def ensemble_select(oof: Dict[str, np.ndarray], Y: np.ndarray, iters: int = 15) -> Dict[str, float]:
    """Greedy forward selection with replacement (Caruana) on out-of-fold EMD.
 
    The constant median curve is one of the candidates, so the selected ensemble can
    never be worse than the trivial solution in cross-validation.
    """
    names = list(oof)
    counts = {n: 0 for n in names}
    cur: Optional[np.ndarray] = None
    best_score = np.inf
    for it in range(iters):
        best_n, best_s = None, np.inf
        for n in names:
            cand = oof[n] if cur is None else (cur * it + oof[n]) / (it + 1)
            s = oof_score(cand, Y)
            if s < best_s - 1e-9:
                best_n, best_s = n, s
        if best_n is None:
            break
        counts[best_n] += 1
        cur = oof[best_n] if cur is None else (cur * it + oof[best_n]) / (it + 1)
        best_score = best_s
    total = sum(counts.values())
    weights = {n: c / total for n, c in counts.items() if c > 0}
    LOG.info("Ensemble weights %s -> OOF EMD %.3f", {k: round(v, 2) for k, v in weights.items()}, best_score)
    return weights
 
 
# ----------------------------------------------------------------------------
# Submission
# ----------------------------------------------------------------------------
def write_submission(ids: Sequence[str], curves: np.ndarray, cols: List[str], path: Path) -> pd.DataFrame:
    sub = pd.DataFrame(np.round(curves, 2), columns=cols)
    sub.insert(0, "sample_id", list(ids))
    sub[cols[-1]] = 100.0
    validate_submission(sub, cols)
    sub.to_csv(path, index=False, float_format="%.2f")
    return sub
 
 
def validate_submission(sub: pd.DataFrame, cols: List[str]) -> None:
    v = sub[cols].values.astype(float)
    assert np.isfinite(v).all(), "non-finite values"
    assert (v >= 0).all() and (v <= 100).all(), "values outside [0, 100]"
    assert (np.diff(v, axis=1) >= -1e-9).all(), "curves are not non-decreasing"
    assert np.allclose(v[:, -1], 100.0), "last column must be 100"
    assert sub["sample_id"].is_unique, "duplicate sample ids"
 
 
# ----------------------------------------------------------------------------
# Main
# ----------------------------------------------------------------------------
def main() -> None:
    out_dir = Path(CFG.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)
    sub_path = out_dir / "submission.csv"
 
    root = resolve_data_dir(CFG.data_dir)
    csvs, imgs = scan_files(root)
    LOG.info("Found %d CSV files and %d images under %s", len(csvs), len(imgs), root)
    files = classify_csvs(csvs)
    if files["train"] is None:
        raise FileNotFoundError("No training CSV with grain size columns found")
 
    train_raw = files["train"][1]
    tid = id_column(train_raw)
    train_raw[tid] = train_raw[tid].astype(str)
    train_df, Y = prepare_targets(train_raw, tid)
 
    # Test ids and submission column layout.
    sub_cols = list(DEFAULT_SUB_COLS)
    test_meta, test_ids = None, None
    if files["sub"] is not None:
        sdf = files["sub"][1]
        cm = support_column_map(sdf.columns)
        sub_cols = [c for c in sorted(cm, key=lambda c: cm[c])]
        test_ids = sdf[id_column(sdf)].astype(str).tolist()
    if files["test"] is not None:
        test_meta = files["test"][1]
        tsid = id_column(test_meta)
        test_meta[tsid] = test_meta[tsid].astype(str)
        if test_ids is None:
            test_ids = test_meta[tsid].tolist()
    if test_ids is None:
        raise FileNotFoundError("Could not determine test sample ids")
    test_sid = id_column(test_meta) if test_meta is not None else None
    if len(sub_cols) != 11:
        sub_cols = list(DEFAULT_SUB_COLS)
 
   
    median_curve = np.median(Y, axis=0)[None, :].repeat(len(test_ids), axis=0)
    write_submission(test_ids, postprocess(median_curve[:, :10]), sub_cols, sub_path)
    LOG.info("Fallback submission written (%d rows)", len(test_ids))
 
    # Image mapping.
    train_imgs = map_images_to_samples(train_df[tid].tolist(), imgs, train_df, tid, root, "train")
    test_imgs = map_images_to_samples(test_ids, imgs, test_meta, test_sid, root, "test")
    n_tr = sum(1 for v in train_imgs.values() if v)
    n_te = sum(1 for v in test_imgs.values() if v)
    LOG.info("Samples with images: train %d/%d, test %d/%d", n_tr, len(train_df), n_te, len(test_ids))
    if n_tr < len(train_df) or n_te < len(test_ids):
        log_mapping_diagnostics("Train", train_df[tid].tolist(), imgs, root, train_df)
        log_mapping_diagnostics("Test", test_ids, imgs, root, test_meta)
    if n_tr < max(5, int(0.3 * len(train_df))):
        raise RuntimeError("Too few training samples could be matched to images; fallback submission kept")
 
    # Features (train and test jointly, then split).
    all_imgs = {**train_imgs, **{k: v for k, v in test_imgs.items() if k not in train_imgs}}
    hand = build_handcrafted(all_imgs)
    deep = build_deep(all_imgs)
    hand.index = hand.index.astype(str)
    if deep is not None:
        deep.index = deep.index.astype(str)
 
    train_ids = train_df[tid].astype(str)
    has_feat = train_ids.isin(hand.index).values
    Y = Y[has_feat]
    train_ids = train_ids[has_feat].reset_index(drop=True)
    train_imgs = {k: v for k, v in train_imgs.items() if k in set(train_ids)}
 
    try:
        run_eda(Y, train_ids, train_imgs, hand, out_dir / "eda")
    except Exception as exc:  # noqa: BLE001
        LOG.info("EDA step failed (%s: %s), continuing", type(exc).__name__, exc)
 
    # Feature groups: 'h' handcrafted, 'd' embeddings.
    groups_tr: Dict[str, np.ndarray] = {}
    groups_te: Dict[str, np.ndarray] = {}
    for gname, frame, thr in (("h", hand, CFG.corr_threshold), ("d", deep, 0.99)):
        if frame is None or frame.empty:
            continue
        cols = filter_features(frame.reindex(train_ids), thr)
        clean = lambda df: df.reindex(columns=cols).replace([np.inf, -np.inf], np.nan).values.astype(np.float32)  # noqa: E731
        groups_tr[gname] = clean(frame.reindex(train_ids))
        groups_te[gname] = clean(frame.reindex(test_ids))
 
    eq = np.tile(np.arange(1, 12) * 100.0 / 11, (len(Y), 1))
    LOG.info("Reference EMD on train labels, equal-share baseline: %.3f", emd_log(Y, eq))
 
    specs: List[ModelSpec] = [("const", "const", "h")]
    for kind in ("knn", "pls", "ridge", "tpca", "et"):
        specs.append((f"{kind}_h", kind, "h"))
    if "d" in groups_tr:
        for kind in ("knn", "ridge", "tpca"):
            specs.append((f"{kind}_d", kind, "d"))
    if CFG.use_lgbm and len(Y) >= 150:
        try:
            import lightgbm  # noqa: F401
            specs.append(("lgbm_h", "lgbm", "h"))
        except ImportError:
            pass
 
    oof = cross_validate(specs, groups_tr, Y)
    if "const" not in oof:
        raise RuntimeError("Constant baseline failed; fallback submission kept")
    table = {n: round(oof_score(p, Y), 3) for n, p in oof.items()}
    LOG.info("OOF EMD per model (repeated CV): %s", dict(sorted(table.items(), key=lambda kv: kv[1])))
    weights = ensemble_select(oof, Y)
 
   
    kind_of = {n: (k, g) for n, k, g in specs}
    pred = np.zeros((len(test_ids), 10))
    for name, w in weights.items():
        kind, grp = kind_of[name]
        model = make_model(kind).fit(groups_tr[grp], Y[:, :10])
        pred += w * model.predict(groups_te[grp])
    no_img = np.array([not test_imgs.get(i) for i in test_ids]) | np.isnan(groups_te["h"]).all(axis=1)
    if no_img.any():
        LOG.info("%d test samples without usable images use the train-median curve", int(no_img.sum()))
        pred[no_img] = np.median(Y, axis=0)[:10]
 
    sub = write_submission(test_ids, postprocess(pred), sub_cols, sub_path)
    LOG.info("Submission saved: %s shape=%s", sub_path, sub.shape)
    LOG.info("Total runtime %.1f min", (time.time() - T0) / 60)
 
 
if __name__ == "__main__":
    try:
        main()
    except Exception as exc:  # noqa: BLE001
        LOG.error("Pipeline error: %s: %s", type(exc).__name__, exc)
        # The early fallback submission (if written) remains valid; do not crash the notebook run.
        if not (Path(CFG.out_dir) / "submission.csv").exists():
            sys.exit(1)
 
```
